# Robert Shiller's stock market data

```python
# load Robert Shiller's stock market data - update link from https://shillerdata.com/
# old link data_Shiller = pd.read_excel('http://www.econ.yale.edu/~shiller/data/ie_data.xls',sheet_name='Data')
# old link 2025: data_Shiller = pd.read_excel('https://img1.wsimg.com/blobby/go/e5e77e0b-59d1-44d9-ab25-4763ac982e53/downloads/9becfac9-1778-47a6-b40e-299d8c616706/ie_data.xls?ver=1756924641552',sheet_name='Data')
data_Shiller = pd.read_excel('https://img1.wsimg.com/blobby/go/e5e77e0b-59d1-44d9-ab25-4763ac982e53/downloads/70fec4f5-727f-4e53-b5f1-179af109c5fa/ie_data.xls?ver=1788371540009',sheet_name='Data')
```

```python
# process the Shiller data - create column names from first rows of the spreadsheet
d = data_Shiller.copy()
d = d.astype('object')
d.iloc[0:7,:] = d.iloc[0:7,:].to_numpy().astype('str')
d[d=='nan'] = ''
d.iloc[0:2,0] = ''
d.columns = d.iloc[0,:]+ ' ' + d.iloc[1,:] + ' ' + d.iloc[2,:] + ' ' + d.iloc[3,:] + ' ' + d.iloc[4,:] + ' ' + d.iloc[5,:] + ' ' + d.iloc[6,:]
d.columns = d.columns.str.strip()       # strip spaces
d = d[7:] 
d = d.drop(d[d['Date'].isna()].index)   # drop rows with NaN Date values
#d.set_index('Date   Fraction');
```

## Average returns

Report average returns over the very long sample.

```python
# Calculate average returns

# stock market
Pe = d['Real Total Return Price'].to_numpy()
t = d['Date   Fraction'].to_numpy()
ERe = ((Pe[-1]/Pe[0])**(1/(t[-1]-t[0]))-1)*100
print(f"Average real annual return on the stock market: {ERe:.3f}%.")
print(f"Average from the 10-year annualized real stock market returns: {d['10 Year Annualized Stock Real Return'].mean()*100:.3f}%.")

# 10-year government bonds
Pb = d['Real Total Bond Returns'].to_numpy()
ERb = ((Pb[-1]/Pb[0])**(1/(t[-1]-t[0]))-1)*100
print(f"Average real annual return on the 10-year bond: {ERb:.3f}%.")
print(f"Average from the 10-year annualized real returns on 10-year government bond: {d['10 Year Annualized Bonds  Real Return'].mean()*100:.3f}%.")
```

```python
param = {'figsize' : [18,9], 'fontsize': 16,
         'title': '',
         'xlim': [1871,graph_end_year], 'ylim': [-6,20],
         'xlabel': 'year (date of investment)', 'ylabel': '(%)',
         'ylogscale': False,
         'showgrid': True, 'highlightzero': True,
         'showNBERrecessions': True, 'showNBERrecessions_y': [-6,20]}

fig,ax = ec.GenerateTSPlot(param)

ax.plot(d['Date   Fraction'],d['10 Year Annualized Stock Real Return']*100,
                    linewidth=3,marker='',color=ec.tolColor['tolVibrantBlue'],label='10-year forward annualized real return on S&P 500')
ax.plot(param['xlim'],[ERe,ERe],
                    linewidth=3,alpha=0.5,linestyle='dotted',marker='',color=ec.tolColor['tolVibrantBlue'])

ax.plot(d['Date   Fraction'],d['10 Year Annualized Bonds  Real Return']*100,
                    linewidth=3,marker='',color=ec.tolColor['tolVibrantOrange'],label='10-year forward annualized real return on 10-year bonds')
ax.plot(param['xlim'],[ERb,ERb],
                    linewidth=3,alpha=0.5,linestyle='dotted',marker='',color=ec.tolColor['tolVibrantOrange'])

x = ax.legend(loc='upper left')

fig.savefig('graphs/17-investment_shiller_forward_real_returns.pdf', bbox_inches='tight')
```

# Example 2: The COVID-19 epidemic

This example uses a processed spreadsheet that includes Robert Shiller's data as well as epidemiological data on reported covid-19 cases from Johns Hopkins University database.

```python
# load a local dataset containing the stock market and epidemic data
data_covid_FRED = pd.read_excel('data/data - covid-19 epidemic.xlsx',sheet_name='FRED daily data')
data_covid_epi = pd.read_excel('data/data - covid-19 epidemic.xlsx',sheet_name='epidemic data')
data_covid_events = pd.read_excel('data/data - covid-19 epidemic.xlsx',sheet_name='events')
```

```python
param = {'figsize' : [18,9], 'fontsize': 16,
         'title': '',
         'xlim': [datetime.datetime(2020,1,1),datetime.datetime(2020,6,16)], 'ylim': [2200,3500],
         'xlabel': 'date', 'ylabel': 'nominal index',
         'ylogscale': False,
         'showgrid': True, 'highlightzero': False,
         'showNBERrecessions': False, 'showNBERrecessions_y': [600,4800]}

dx = data_covid_FRED['day'].to_numpy()
dy = data_covid_FRED['S&P 500 close'].to_numpy()

fig,ax1 = ec.GenerateTSPlot(param)

ln1 = ax1.plot(data_covid_FRED['day'][np.isnan(dy)==False],dy[np.isnan(dy)==False],
                    linewidth=3,marker='',color=ec.tolColor['tolVibrantBlue'],label='S&P 500 nominal index (left axis)')
#ax1.set_xlim([datetime.datetime(2020,1,1),datetime.datetime(2020,6,16)])

# add right axis with epidemic data
ax2 = ax1.twinx()  # instantiate a second axes that shares the same x-axis
ln2 = ax2.plot(data_covid_epi['day'],data_covid_epi['China new cases'],
                    linewidth=1,marker='',color=ec.tolColor['tolVibrantOrange'],label='China new cases (right axis)',alpha=0.5)
ln3 = ax2.plot(data_covid_epi['day'],data_covid_epi['Italy new cases'],
                    linewidth=1,marker='',color=ec.tolColor['tolVibrantTeal'],label='Italy new cases (right axis)',alpha=0.5)
ln4 = ax2.plot(data_covid_epi['day'],data_covid_epi['New York City new cases'],
                    linewidth=1,marker='',color=ec.tolColor['tolVibrantMagenta'],label='New York City new cases (right axis)',alpha=0.5)
ax2.set_ylim([0,8000])
ax2.set_ylabel('daily new cases')

# add legend
lns = ln1 + ln2 + ln3 + ln4
labs = [l.get_label() for l in lns]
ax1.legend(lns, labs, loc='upper right')

# roll through all events
for index,row in data_covid_events.iterrows():
    
    fig,ax1 = ec.GenerateTSPlot(param)

    ln1 = ax1.plot(data_covid_FRED['day'][np.isnan(dy)==False],dy[np.isnan(dy)==False],
                        linewidth=3,marker='',color=ec.tolColor['tolVibrantBlue'],label='S&P 500 nominal index (left axis)')
    ax1.set_xlim([datetime.datetime(2020,1,1),datetime.datetime(2020,6,16)])

    # add right axis with epidemic data
    ax2 = ax1.twinx()  # instantiate a second axes that shares the same x-axis
    ln2 = ax2.plot(data_covid_epi['day'],data_covid_epi['China new cases'],
                        linewidth=1,marker='',color=ec.tolColor['tolVibrantOrange'],label='China new cases (right axis)',alpha=0.5)
    ln3 = ax2.plot(data_covid_epi['day'],data_covid_epi['Italy new cases'],
                        linewidth=1,marker='',color=ec.tolColor['tolVibrantTeal'],label='Italy new cases (right axis)',alpha=0.5)
    ln4 = ax2.plot(data_covid_epi['day'],data_covid_epi['New York City new cases'],
                        linewidth=1,marker='',color=ec.tolColor['tolVibrantMagenta'],label='New York City new cases (right axis)',alpha=0.5)
    ax2.set_ylim([0,8000])
    ax2.set_ylabel('daily new cases')

    ax1.plot([row['day'],row['day']],param['ylim'],linewidth=3,linestyle='dotted',color=ec.tolColor['tolHighContrastBlack'])
    event_text = row['event'].replace('|','\n')
    if ((row['day']-param['xlim'][0])/(param['xlim'][1]-param['xlim'][0]))<0.5:        
        t = ax1.text(row['day']+datetime.timedelta(days=2),param['ylim'][1]*0.99,event_text,
                 fontsize=20,va='top')
    else:
        t = ax1.text(row['day']-datetime.timedelta(days=2),param['ylim'][1]*0.99,event_text,
                 fontsize=20,va='top',ha='right')
    tbox = t.get_bbox_patch()
    t.set_bbox(dict(facecolor=ec.tolColor['tolHighContrastYellow'], alpha=1, edgecolor=ec.tolColor['tolHighContrastYellow']))

    fig.savefig(f'graphs/17-investment_shiller_SP500_covid_episode_{index+1:02d}.pdf', bbox_inches='tight')
```
