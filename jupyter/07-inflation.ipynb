## Ex-post real rate

```python
# ex-post real rate computed from nominal rate and realized inflation rate
data = ec.LoadDataFRED(series=['TB3MS','GS1','CPIAUCSL'],transform='none')
period_shift = 12

#data['ex_post_real_rate'] = data['TB3MS'] - data['CPIAUCSL'].pct_change(periods=12)*100

data['forward inflation'] = ((data['orig']['CPIAUCSL'].shift(-period_shift) / data['orig']['CPIAUCSL'])**(12/period_shift) - 1)*100

param = {'figsize' : [15,9], 'fontsize': 18,
         'title': '',
         'xlim': [1955,graph_end_year], 'ylim': [-10,20],
         'xlabel': 'year', 'ylabel': 'Percent',
         'ylogscale': False,
         'showgrid': True, 'highlightzero': True,
         'showNBERrecessions': True, 'showNBERrecessions_y': [-10,20]}

fig,ax = ec.GenerateTSPlot(param)

ax.plot(data['year'],data['GS1'],
                    '--',linewidth=3,marker='',color=ec.tolColor['tolVibrantOrange'],label='1-Year Treasury Constant Maturity Rate')
ax.plot(data['year'],data['GS1'] - data['forward inflation'],
                    '-',linewidth=3,marker='',color=ec.tolColor['tolVibrantBlue'],label='Ex-post real interest rate (1-year nominal rate $-$ realized inflation)')
x = ax.legend(loc='lower left')
#ax.set_xticks([2002,2004,2006,2008,2010,2012,2014,2016,2018,2020,2022,2024,2026]);

ax.annotate(f'Loose monetary policy',
    xy=(1980,-5), xycoords="data", xytext=(60, 0), textcoords="offset points", verticalalignment='center', horizontalalignment='left', fontsize=18, backgroundcolor=ec.tolColor['tolHighContrastWhite'],
    arrowprops=dict(arrowstyle="simple,head_length=1.4,head_width=1.8,tail_width=1.0", relpos=(0,0.5), shrinkA=10,shrinkB=0,
                    facecolor=ec.tolColor['tolVibrantRed'], edgecolor=ec.tolColor['tolVibrantBlue'], linewidth=1.5,joinstyle='miter'),zorder=12)

ax.annotate(f'Volcker tightening',
    xy=(1982,3), xycoords="data", xytext=(0, -70), textcoords="offset points", verticalalignment='center', horizontalalignment='left', fontsize=18, backgroundcolor=ec.tolColor['tolHighContrastWhite'],
    arrowprops=dict(arrowstyle="simple,head_length=1.4,head_width=1.8,tail_width=1.0", relpos=(0,1), shrinkA=10,shrinkB=0,
                    facecolor=ec.tolColor['tolVibrantRed'], edgecolor=ec.tolColor['tolVibrantBlue'], linewidth=1.5,joinstyle='miter'),zorder=12)


fig.savefig('graphs/08-inflation_real_interest_rate_ex_post.pdf', bbox_inches='tight')
```
